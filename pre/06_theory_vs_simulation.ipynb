# 06. Theory against simulation: spinodals, the alpha = 0 clock, and rare outside patterns

**Story.** A memory that rewrites itself needs outside data. Mean field gives two edges:
- the **diverse** state (both modes have a basin) dies below alpha_sym(beta);
- the **consensus** state (all self-written patterns in one mode) dies above alpha_cons(beta), the Curie-Weiss spinodal.

Between them both states can live (first order). At sharp retrieval one outside pattern from the tail of the environment can found its own basin, because its weight grows like e^{beta r (r+1)} with its norm r.

*Example.* beta = 4, K = 1280: a single outside pattern with norm 1.24 (2.4 sigma) is a critical nucleus. A unit-norm nucleus would need 18 patterns.
